## Key Findings

- The FinanceBench sample contains 150 question-answer examples across 32 companies and 84 financial documents.
- The dataset is evenly split across three question types: metrics-generated, domain-relevant, and novel-generated.
- Questions require a mix of information extraction, numerical reasoning, and logical reasoning.
- Most source documents are 10-K filings, with additional 10-Q, earnings, and 8-K documents.
- Each question includes grounded evidence from the underlying financial document, including page-level metadata.
- Some questions require multiple evidence passages, which is important for retrieval evaluation.
- There are no duplicate questions.
- Some metadata fields are intentionally missing depending on the question type.

## Test Split Consideration

Multiple questions can come from the same source document. Therefore, a purely random question-level train/test split could place questions from the same filing in both sets.

For a more realistic evaluation, we split the data by `doc_name`, so that all questions from the same financial document remain in the same partition. This would reduce leakage and better test whether the RAG system can retrieve and answer questions from unseen documents.

```python
from sklearn.model_selection import train_test_split

docs = df["doc_name"].unique().tolist()

train_docs, test_docs = train_test_split(
    docs,
    test_size=0.2,
    random_state=42
)

train_df = df[df["doc_name"].isin(train_docs)]
test_df = df[df["doc_name"].isin(test_docs)]

print("Train questions:", len(train_df))
print("Test questions:", len(test_df))

shared_docs = set(train_df["doc_name"]) & set(test_df["doc_name"])
print("Shared documents:", shared_docs)
```
